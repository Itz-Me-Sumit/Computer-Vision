```python
import cv2
image = cv2.imread("sumit.jpg")
print(image.shape)
```

```python
# Convery our image to grayscale

gray_image = cv2.cvtColor(image , cv2.COLOR_BGR2GRAY)

import matplotlib.pyplot as plt
plt.imshow(gray_image , cmap="gray")
plt.axis("off")
plt.show()
```
