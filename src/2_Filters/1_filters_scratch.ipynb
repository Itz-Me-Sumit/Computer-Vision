```python
import torch
```

```python
img = torch.tensor([
    [1, 2, 3, 0, 1],
    [4, 5, 6, 1, 0],
    [7, 8, 9, 2, 1],
    [1, 0, 2, 3, 4],
    [0, 1, 3, 5, 6],
], dtype = torch.float32)

kernel = torch.tensor([
    [1 , 0 , -1],
    [1 , 0 , -1],
    [1 , 0 , -1]
])


k = 3 # Size of kernel
out_size = img.shape[0] - k + 1
out = torch.zeros(out_size , out_size)

for i in range(out_size):
    for j in range(out_size):
        patch = img[ i : i+k , j : j+k ]
        out[i , j] = (patch * kernel).sum()

print(out)
```
